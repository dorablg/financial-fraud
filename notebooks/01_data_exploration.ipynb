# 01 Data exploration

Shape, fraud rate, time span, missing values and amounts for the IEEE-CIS train data.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_parquet('../data/interim/train.parquet')  # created by: python src/load.py
df.shape
```

```python
print('fraud rate', df['isFraud'].mean())
print('days', df['TransactionDT'].max() / 86400)
```

```python
missing = df.isna().mean().sort_values(ascending=False)
missing.head(10)
```

```python
df['day'] = df['TransactionDT'] // 86400
df.groupby(df['day'] // 7)['isFraud'].mean().plot(title='Fraud rate per week')
plt.show()
```
